## Partie 1 – Series

Une **Series** est la structure de base de pandas : une colonne de données
à une seule dimension. Chaque valeur possède une étiquette appelée **index**.
On peut la voir comme une seule colonne d'un tableur, où l'index joue le rôle
des numéros de ligne.

### 1. Créer une Series contenant 4 températures

- **Objectif** : créer une Series à partir d'une liste de valeurs.
- **Syntaxe** : `pd.Series([valeur1, valeur2, ...])`

```python
import pandas as pd

temperatures = pd.Series([22.5, 24.1, 19.8, 26.3])
temperatures
```

### 2. Afficher l'index de la Series

- **Objectif** : voir les étiquettes attribuées à chaque valeur.
- **Syntaxe** : `ma_series.index`

```python
temperatures.index
```

### 3 & 4. Afficher la première et la dernière valeur

- **Objectif** : accéder à une valeur par sa **position**.
- **Syntaxe** : `ma_series.iloc[position]`
  - `iloc` = *integer location* = accès par position (0 = première).
  - La position `-1` désigne toujours le **dernier** élément.

```python
print("Première valeur :", temperatures.iloc[0])
print("Dernière valeur :", temperatures.iloc[-1])
```

### 5. Créer une Series de températures indexée par les heures (12h à 15h)

Deux méthodes possibles :
- **Objectif** : associer à chaque température une étiquette (l'heure de mesure).
- **Syntaxe 1** (paramètre `index`) : `pd.Series(valeurs, index=liste_etiquettes)`
- **Syntaxe 2** (dictionnaire) : `pd.Series({etiquette: valeur, ...})`
  → les **clés** du dictionnaire deviennent automatiquement l'index.

```python
# Façon 1 : avec le paramètre index
temp_heures_1 = pd.Series([22.5, 24.1, 19.8, 26.3], index=["12h", "13h", "14h", "15h"])
print("Façon 1 :")
print(temp_heures_1)

# Façon 2 : à partir d'un dictionnaire
temp_heures_2 = pd.Series({"12h": 22.5, "13h": 24.1, "14h": 19.8, "15h": 26.3})
print("\nFaçon 2 :")
print(temp_heures_2)
```

### 6. Afficher la première et la dernière valeur

- **Objectif** : réutiliser l'accès par position sur la Series indexée par heures.
- **Syntaxe** : `ma_series.iloc[0]` et `ma_series.iloc[-1]`
  (On pourrait aussi passer par l'étiquette : `ma_series["12h"]`, `ma_series["15h"]`.)

```python
print("Première valeur :", temp_heures_1.iloc[0])
print("Dernière valeur :", temp_heures_1.iloc[-1])
```

## Partie 2 – DataFrame

Un **DataFrame** est un tableau à deux dimensions : plusieurs colonnes nommées
et des lignes indexées. C'est la structure centrale de pandas, équivalente à
une feuille de calcul.

### 1. Créer et afficher le DataFrame df_test

- **Objectif** : construire un DataFrame à partir d'un dictionnaire.
- **Syntaxe** : `pd.DataFrame(dictionnaire)`
  → chaque **clé** du dictionnaire devient un **nom de colonne**,
    et la **liste** associée devient le **contenu** de la colonne.

```python
data = {
    "capteur": ["C001", "C002", "C003"],
    "temperature": [24.5, 25.1, 26.3],
    "humidite": [65, 67, 70]
}

df_test = pd.DataFrame(data)
df_test
```

### 2. Importer le dataset mesures_capteurs.csv dans le DataFrame df

- **Objectif** : charger un fichier CSV pour le transformer en DataFrame.
- **Syntaxe** : `pd.read_csv("chemin/vers/le/fichier.csv")`
  → `read_csv` lit un fichier de valeurs séparées par des virgules
    (CSV = *Comma-Separated Values*) et le convertit en DataFrame.

```python
df = pd.read_csv("../data/mesures_capteurs.csv")
df
```

## Partie 4 – Sélection

Sélectionner, c'est extraire une partie du tableau : une colonne, plusieurs
colonnes, ou un sous-ensemble de lignes. Deux outils clés : `loc` (par étiquette)
et `iloc` (par position).

### 1. Sélectionner la colonne "temperature"
- **Objectif** : extraire une seule colonne (renvoie une Series).
- **Syntaxe** : `df["nom_colonne"]`

```python
df["temperature"]
```

### 2. Sélectionner plusieurs colonnes
- **Objectif** : extraire plusieurs colonnes (renvoie un DataFrame).
- **Syntaxe** : `df[["col1", "col2", ...]]` → noter les **doubles crochets**.

```python
df[["temperature", "humidite", "pression", "consommation"]]
```

### 3. Les 5 premières lignes avec loc
- **Objectif** : sélectionner des lignes par leur étiquette d'index.
- **Syntaxe** : `df.loc[debut:fin]` → avec `loc`, la borne de fin est **incluse**.

```python
df.loc[0:4]
```

### 4. Les 5 premières lignes et 4 premières colonnes avec iloc
- **Objectif** : sélectionner par **position** (numéros), lignes ET colonnes.
- **Syntaxe** : `df.iloc[lignes, colonnes]` → avec `iloc`, la borne de fin est **exclue**.

```python
df.iloc[0:5, 0:4]
```

## Partie 5 – Manipulation des colonnes

On peut créer de nouvelles colonnes, les catégoriser, les renommer ou les
supprimer. Ces opérations préparent les données pour l'analyse.

### 1. Ajouter la colonne "temperature_fahrenheit"
- **Objectif** : créer une colonne calculée à partir d'une autre.
- **Syntaxe** : `df["nouvelle"] = <calcul sur une colonne existante>`
- **Formule** : °F = °C × 9/5 + 32

```python
df["temperature_fahrenheit"] = df["temperature"] * 9/5 + 32
df[["temperature", "temperature_fahrenheit"]].head()
```

### 2. Ajouter la colonne "niveau_temperature"
- **Objectif** : catégoriser chaque ligne selon une condition.
- **Syntaxe** : `np.where(condition, valeur_si_vrai, valeur_si_faux)`
  → il faut importer numpy : `import numpy as np`.

```python
import numpy as np

df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df["niveau_temperature"].value_counts()
```

### 3. Renommer "humidite" en "humidite_relative"
- **Objectif** : changer le nom d'une colonne.
- **Syntaxe** : `df.rename(columns={"ancien": "nouveau"})`
  → on réassigne le résultat à `df` pour que le changement soit conservé.

```python
df = df.rename(columns={"humidite": "humidite_relative"})
df.columns
```

### 4. Supprimer la colonne "temperature_fahrenheit"
- **Objectif** : retirer une colonne devenue inutile.
- **Syntaxe** : `df.drop(columns=["nom_colonne"])`

```python
df = df.drop(columns=["temperature_fahrenheit"])
df.columns
```
